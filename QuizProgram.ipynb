# Simple Quiz Game

The file "QuizData.txt" should be in the same folder as this program.

```python
def stripLastNewline(s):
    'Returns the string with the ending newline removed'
    if s[-1] == '\n':
        return s[:-1]
    else:
        return s
```

```python
data = open("QuizData.txt", "r")

num_correct = 0
total = 0
while True:
    question = data.readline()
    answer   = data.readline()
    
    if question == '':
        break
        
    question = stripLastNewline(question)
    answer   = stripLastNewline(answer)
    response = input(question)
    if response.lower() == answer.lower():
        print("Correct!")
        num_correct += 1
    else:
        print("Wrong!")
    
    total += 1
    print()
    
print("Percent correct:", 100*num_correct / total)

data.close()
```
